# EdgeFireUAV — Dual Model Independent Inference & Visual Comparison

This notebook provides a complete testing suite for both independent detection models:
- **Model A**: Fire & Smoke Detector (`smoke`, `fire`)
- **Model B**: 3-Class Custom Posture Detector (`Lying`, `Sitting`, `Standing`)

Tested on `test.jpeg` and `test2.jpg`.

```python
import os
from pathlib import Path
import cv2
import numpy as np
from PIL import Image
from IPython.display import display
from ultralytics import YOLO

project_dir = Path(os.getcwd())
test_images = [(project_dir / "test_media" / f) if (project_dir / "test_media" / f).exists() else (project_dir / f) for f in ["test.jpeg", "test2.jpg"]]

print("Project Directory:", project_dir)
for img in test_images:
    print(f"Test Image '{img.name}': Exists = {img.exists()}")
```

## 1. Load Both Independent Models

```python
# Define candidate weight paths for both models
fire_weights = [
    project_dir / "models/training/fire_detector.onnx",
    project_dir / "models/training/runs/fire_detector/train/weights/best.pt",
    "yolov8n.pt"
]

posture_weights = [
    project_dir / "models/training/custom_posture_detector.onnx",
    project_dir / "models/training/runs/custom_posture_detector/train/weights/best.pt",
    project_dir / "runs/detect/runs/custom_posture_detector/train/weights/best.pt",
    project_dir / "runs/detect/runs/custom_posture_detector/train/weights/best.pt",
    "yolov8n.pt"
]

fire_model_path = str(next((w for w in fire_weights if isinstance(w, str) or w.exists()), "yolov8n.pt"))
posture_model_path = str(next((w for w in posture_weights if isinstance(w, str) or w.exists()), "yolov8n.pt"))

print(f"[Model 1] Fire Detector Weight: {fire_model_path}")
fire_model = YOLO(fire_model_path)

print(f"[Model 2] Custom Posture Detector Weight: {posture_model_path}")
posture_model = YOLO(posture_model_path)

print("\nFire Model Classes:", fire_model.names)
print("Posture Model Classes:", posture_model.names)
```

## 2. Independent Model Inferences on Test Images

```python
for img_path in test_images:
    if not img_path.exists():
        continue
    
    print(f"\n{'='*70}")
    print(f"  TESTING IMAGE: {img_path.name}")
    print(f"{'='*70}")
    
    # Fire Model Inference
    fire_res = fire_model.predict(source=str(img_path), conf=0.25)[0]
    fire_img = Image.fromarray(fire_res.plot()[..., ::-1])
    
    # Posture Model Inference
    posture_res = posture_model.predict(source=str(img_path), conf=0.35, classes=[0] if posture_model_path == 'yolov8n.pt' else None)[0]
    posture_img = Image.fromarray(posture_res.plot()[..., ::-1])
    
    # Print Detection Dictionaries
    print("\n[Fire Detector Detections]")
    for box in fire_res.boxes:
        cls_id = int(box.cls[0])
        cls_name = fire_model.names.get(cls_id, str(cls_id))
        conf = float(box.conf[0])
        print(f"  - {cls_name}: confidence = {conf:.2f}")
    if len(fire_res.boxes) == 0:
        print("  (No fire/smoke detected above confidence threshold)")
        
    print("\n[Custom Posture Detector Detections]")
    for box in posture_res.boxes:
        cls_id = int(box.cls[0])
        cls_name = posture_model.names.get(cls_id, str(cls_id))
        conf = float(box.conf[0])
        print(f"  - {cls_name}: confidence = {conf:.2f}")
    if len(posture_res.boxes) == 0:
        print("  (No human posture detected above confidence threshold)")
    
    # Display annotated images side by side
    print("\nVisual Overlay (Left: Fire Detector | Right: Custom Posture Detector):")
    canvas = np.hstack([np.array(fire_img), np.array(posture_img)])
    display(Image.fromarray(canvas))
```
